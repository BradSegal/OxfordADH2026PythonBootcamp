## A quick recap

Python uses standard order of operations as you're used to, i.e., brackets,
then `**` (powers), then `*` & `/`, then `+` & `-`. A variable is defined
using `=` which is the assignment operator, i.e., it means 'store the
following at this name' and not 'is equal to'.

````python
weight_kg = 81.4
height_m = 1.72
bmi = weight_kg / height_m**2
print(bmi)
print(round(bmi, 1))
````

## Problem 1.5: Fix the bug

This calculation runs without an error, but the BMI it prints is wrong.
**There are two mistakes**: find and fix both. Height is recorded in
centimetres, but the BMI formula needs metres.

````python
weight_kg = 81.4
height_cm = 172
height_m = height_cm / 100
bmi = weight_kg / height_m**2
print(f"BMI: {bmi:.1f}")
check.bmi_from_cm(weight_kg, height_cm, bmi)
````

Two bugs! The `* 2` is doubling the number when we want `** 2` (squared) and
the height needs to be in metres. Just because the code runs without error
doesn't mean it's right. You should always think if the answer makes
clinical sense. If you get a BMI of ± 0.9, 0.003, or 95 you should probably
stop and relook.

## Problem 1.6: Make, updating a variable

A running fluid balance is tracked through the shift. Start with `balance_ml
= 0`. Add each intake and subtract each output **in order below**, one per
line using `+=` and `-=`:

| Time  | Event           | Volume (ml) |
|-------|-----------------|-------------|
| 08:00 | IV fluids in    | 500         |
| 10:00 | Oral intake     | 250         |
| 12:00 | Urine out       | 400         |
| 14:00 | IV fluids in    | 700         |
| 16:00 | Urine out       | 700         |

`balance_ml += 500` is shorthand for saying `balance_ml = balance_ml + 500`.

````python
balance_ml = 0
balance_ml += 500
balance_ml += 250
balance_ml -= 400
balance_ml += 700
balance_ml -= 700
print(f"Fluid balance: {balance_ml} ml")
check.fluid_balance(balance_ml)
````

## Before you move on

- `=` is used to store a value, not to check equality.
- Brackets explicitly indicate the order of operations. If in doubt, use
  them.
- A result which runs but is clinically implausible is still a bug.
