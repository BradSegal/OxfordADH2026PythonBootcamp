## Comparisons give True or False

A **comparison** is a question asked about two values and results in a
boolean answer.

| Operator | Question |
|---|---|
| `<` | less than? |
| `<=` | less than or equal to? |
| `>` | greater than? |
| `>=` | greater than or equal to? |
| `==` | equal to? |
| `!=` | not equal to? |

Note the double equals here. A single `=` saves a value (Level 1), however,
`==` asks if two values are the same. This is a very common mistake.

````python
pulse = 96
print(pulse > 90)
print(pulse == 96)
print(pulse != 96)
print(pulse < 96)
print(pulse <= 96)

consciousness = "A"
print(consciousness == "A")
````

## if, elif and else

`if` runs some lines only when a condition is `True`. The lines that belong
to it are **indented** (moved in by four spaces), and the `if` line ends in
a colon. Indentation is how Python knows where the `if` block ends.

`else` provides code for if none of the above are `True`. `elif` (Else if)
provides additional conditions to check. Python evaluates these in order
from top to bottom and executes **only the first** code block that is True
and then skip the rest.

````python
consciousness = "V"

if consciousness == "A":
    print("Alert")
elif consciousness == "C":
    print("New confusion")
elif consciousness == "V":
    print("Responds to voice")
else:
    print("Responds to pain, or unresponsive")

print("This line is not indented, so it always runs.")
````

Change the `"V"` to an `"A"`, `"C"`, or `"U"` and run the cell again to see
a different branch run.

## Why the next problems sit inside a function

A rule is only right if it is right for **every** value and especially at
the edges between bands. To test that, the checker needs to run your `if`
block many times with different values. So, from here on your `if` block is
contained within a small **function** as you saw in the basics notebook.
This is a named piece of code that takes a value in and gives an answer
back.

```python
def bmi_category(bmi):        # the name, and the value it takes in
    ...                       # your if block goes here, indented
    return category           # the answer it gives back
```

You only write the indented lines in the middle, the `def` line and the
`return` line are done for you. Everything inside the function is indented
by four spaces and each `if` block by another four. Level 6 shows you how to
write functions from scratch.

## Problem 3.2: Modify

The adult BMI bands (WHO) are:

| BMI | Category |
|---|---|
| below 18.5 | underweight |
| 18.5 to below 25 | healthy |
| 25 to below 30 | overweight |
| 30 or above | obese |

These WHO bands are the general ones. NICE recommends lower thresholds (23
for overweight, 27.5 for obesity) for people from South Asian, Chinese,
other Asian, Middle Eastern, Black African or African-Caribbean family
backgrounds; however, this question uses the general bands only.

The function below is missing one of the four bands and so calls a BMI of 41
'overweight'. Fix this to include the obese band. Make sure the names match
the table exactly and remain lower case.

````python
def bmi_category(bmi):
    if bmi < 18.5:
        category = "underweight"
    elif bmi < 25:
        category = "healthy"
    elif bmi < 30:
        category = "overweight"
    else:
        category = "obese"
    return category


print(bmi_category(27.5))
print(bmi_category(41.0))
check.bmi_category(bmi_category)
````

## Boundaries: < or <=?

## Problem 3.3: Make

Write the NEWS2 temperature score. These bands are from the Royal College of
Physicians NEWS2 chart:

| Temperature (°C) | Score |
|---|---|
| 35.0 or below | 3 |

Replace the `score = ...` line with an `if`, `elif`, and `else` block that

## and, or, not

Clinical rules often depend on more than one thing. Booleans can be combined
with three words:

- `a and b` is `True` only when **both** are `True`;
- `a or b` is `True` when **at least one** is `True`;
- `not a` flips `True` to `False` and back.

One trap catches almost everyone. Each side of `and` or `or` must be a
complete comparison. Write `consciousness == "V" or consciousness == "P"`,
not `consciousness == "V" or "P"`. The second version never compares
`"P"` with anything: Python treats `"P"` on its own as true, so the
condition passes for every patient.

````python
on_oxygen = True
consciousness = "A"
print(on_oxygen and consciousness == "A")  # True: both are True
print(on_oxygen and consciousness == "V")  # False: the second is False
print(on_oxygen or consciousness == "V")  # True: at least one is True
print(not on_oxygen)  # False: not flips the value

alert_on_oxygen = on_oxygen and consciousness == "A"  # store the answer
print(alert_on_oxygen)
````

## Problem 3.4: Make, with or

On the NEWS2 chart a score of 3 in any single parameter is called a **red
score**. It calls for an urgent review even when the total score is low.
A 3 in any parameter counts, including blood pressure, consciousness and
temperature; the supplemental oxygen item can only reach 2. To keep the
function short, this problem looks at three parameters only.

Write `has_red_score` so that `red` is `True` if any of the respiration,
SpO2, or pulse scores are 3, else `False`. This can be done in a single line
with an `or` statement as the comparison evaluates to `True` or `False`
which can be directly saved.

````python
def has_red_score(respiration_score, spo2_score, pulse_score):
    red = respiration_score == 3 or spo2_score == 3 or pulse_score == 3
    return red


print(has_red_score(0, 3, 1))
check.has_red_score(has_red_score)
````

## Problem 3.6: Fix the bug

This function should give the NEWS2 SpO2 score if on Scale 1 which is the
scale most patients are on:

| SpO2 (%) | Score |
|---|---|
| 91 or below | 3 |
| 92 to 93 | 2 |
| 94 to 95 | 1 |
| 96 or above | 0 |

It runs without an error, however, if a patient has saturations of 86% they
score 1. Every condition matches the chart but there is a bug, it's just not
on any one line.

Find and fix the bug.

````python
def score_spo2(spo2):
    if spo2 <= 91:
        score = 3
    elif spo2 <= 93:
        score = 2
    elif spo2 <= 95:
        score = 1
    else:
        score = 0
    return score


print(score_spo2(86))
check.score_spo2(score_spo2)
````

Python short-circuits at the first `True` condition. Any saturation less
than or equal to 93 is also less than or equal to 95, so the `spo2 <= 95`
condition captured all low values first and so the bottom two branches would
never be reached. If using `<=` for banding, ensure you check from the
lowest band first (or highest if using `>=`). A rule may be correct per line
but incorrect overall.
