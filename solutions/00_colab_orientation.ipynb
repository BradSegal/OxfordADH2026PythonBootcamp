## Order matters

Python remembers what you've run **in the order you ran it**, not
necessarily the order the cells appear on the page. Try running the
following three cells in order, from top to bottom.

````python
patients_waiting = 4
````

````python
print(patients_waiting)
````

````python
patients_waiting = 9
````

The middle cell printed 4. Now go back up and run the middle cell again.
This time it prints 9, because the last thing you ran set
`patients_waiting` to 9, even though the cell just above still says 4.

This is the most common source of issues with notebooks. If something is
behaving oddly, the solution is typically to run the cells again in order
from the top.

## Restarting

Sometimes you want to start fresh. **Runtime > Restart session** makes Colab
forget everything it has run (in older versions of Colab this is called
*Restart runtime*). Your code and text stay on the page, but Python's memory
is cleared.

Colab also restarts by itself if you leave a notebook idle for a while.
After any restart, run the cells again from the top. Two shortcuts help:

- **Runtime > Run all** runs every cell in order.
- **Runtime > Run before** runs every cell above the one you have selected.

Try it now. Click **Runtime > Restart session** and then run just the
`print(patients_waiting)` cell above. You will get an error because Python
doesn't know what `patients_waiting` is as it was never told after the

## Reading an error message

Errors are normal. Everyone who writes code experiences them every day and
the message usually gives you a clear indication of what went wrong. The
error you just saw looks something like this (the details in the middle
differ):

```
---------------------------------------------------------------------------
NameError                                 Traceback (most recent call last)
<ipython-input-4-0b8e3a1c2f7d> in <cell line: 0>()
----> 1 print(patients_waiting)

NameError: name 'patients_waiting' is not defined
```

This whole message is called a **traceback**. You read it in this order:

1. **The last line first.** The word before the colon is the type of error
(`NameError`). After the colon, Python gives a sentence explanation that
`patients_waiting` is not defined.
2. **Then the arrow.** `---->` points at the line in your code that Python
   stopped.

A `NameError` means that Python does not know what a particular name is.
This is commonly due to a typo (`patient_waiting` instead of
`patients_waiting`) or as in this case, the cell that produces the variable
name has not been run since the last restart.

Colab may also give you a button offering to explain the error with AI. Try
reading the last line yourself first. It's usually faster and the skill you
will need.

### A second error to read

The below cell has a deliberate error in it but is disabled by the `#` at
the start of the line. The `#` causes the rest of the line to be treated as
a **comment** which Python ignores.

Delete the `#` and space so that the line starts with `print`
and run the cell. Look at the last line of the error.

````python
# print("Bed " + 12)
````

The last line reads:

```
TypeError: can only concatenate str (not "int") to str
```

`str` is Python's short name for text and `int` for a whole number;
"concatenate" means join. Text in quotes, like `"Bed "`, and a number, like
`12`, are different kinds of value. `+` can join two pieces of text or add
two numbers, but Python will not guess what you mean when you mix them, so
it stops with an error.

Fix the line by putting quotes around the 12, so it reads
`print("Bed " + "12")`, and run the cell again. Now both pieces are text,
and it prints `Bed 12`.
