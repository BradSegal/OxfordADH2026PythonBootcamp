## 1. Variables: the facts of the case

A **variable** is a name for a value you want to keep. You create one with
`=`, which in Python means "store this value under this name". It does not
mean "is equal to". Once stored, you can use the name wherever you need the
value.

````python
age = 54
bed = 12
print(age)
print(bed)
````

Good names make code readable, the same way that clear documentation makes a
chart readable. By convention in Python, names are written with lower case
letters with underscores between words. A name cannot contain spaces. If you
include the unit within the name, such as `weight_kg`, the next person to
read your code doesn't have to guess what you were thinking.

### Numbers

Python has two forms of numbers. These are whole numbers known as
**integers**
(`54`) and decimal numbers known as **floats** (`81.4`). Arithmetic can be
performed using standard symbols of `+`, `-`, `*` (multiply), `/` (divide),
and
`**` (to the power of). Here is Jordan's body mass index (BMI) by
calculating
the weight over the height squared:

````python
weight_kg = 81.4
height_m = 1.72
bmi = weight_kg / height_m**2
print(bmi)
print(round(bmi, 1))
````

`round(bmi, 1)` rounds to one decimal place. `round` and `print` are
**functions**: named actions that you call by writing the name followed by
brackets. Whatever goes inside the brackets is the information the function
works on. You will write your own functions in section 5.

### Text

Text in Python is called a **string**, and goes inside quotes. Double
quotes `"..."` and single quotes `'...'` both work; this course uses double
quotes.

````python
name = "Jordan Nguyen"
ward = "5B"
diagnosis = "Community-acquired pneumonia"
print(name)
print(diagnosis)
````

To construct a sentence from variables you can add an `f` just before the
opening quotation mark and place the variable name within curly brackets
`{}`. This is called an **f-string** (the f stands for 'formatted') and lets
Python know to replace the indicated `{name}` with the value stored under
the name.

````python
summary = f"{name}, aged {age}. Diagnosis: {diagnosis}"
print(summary)
````

The text not in the curly brackets such as `, aged ` is kept exactly as you
typed it including spaces and punctuation. Level 2 goes through strings and
f-strings in depth.

## 2. Lists: following a trend

A single value isn't very informative but seeing the trend is far more
useful. A **list** is a collection of values in sequence. It is defined with
square brackets and comma-separated values. Here are the results from
Jordan's seven rounds over the course of the last day from oldest to latest
from their chart:

````python
times = ["06:00", "10:00", "14:00", "18:00", "22:00", "02:00", "06:00"]
pulse_readings = [112, 106, 101, 96, 92, 88, 84]
temperatures = [38.6, 38.3, 37.9, 37.6, 37.3, 37.0, 36.8]
print(pulse_readings)
````

### Picking out one reading

Each element in a list has a number associated with it called an **index**.
Python uses a 0-based indexing system, so the first element in the list is
at index 0. You can also use negative numbers to count backwards from the
end of the list, so the last element in the list is at index -1. To select
an element from a list, you use the name of the list and add a square
bracket with the index number you want to access. The `print` function can
also take multiple inputs and prints them in order if separated by a comma.

````python
print(times[0], pulse_readings[0])  # the first round
print(times[1], pulse_readings[1])  # the second round
print(times[-1], pulse_readings[-1])  # the latest round
````

Text after a `#` is a **comment**: a note for human readers that Python
ignores.

### Summarising a list

Python has functions that can summarise a list of numbers: `len` (how many
items), `max`, `min`, and `sum`. The mean is the sum divided by the count.

````python
print(len(temperatures))
print(max(temperatures))
print(min(temperatures))
mean_temperature = sum(temperatures) / len(temperatures)
print(round(mean_temperature, 1))
````

### Adding a reading

Lists can grow. `.append(value)` adds a value to the end of a list. The
dot means "do this to the list": `times.append("10:00")` adds `"10:00"` to
`times`. An action attached to a value like this is called a **method**.

````python
times = ["06:00", "10:00", "14:00", "18:00", "22:00", "02:00", "06:00"]
times.append("10:00")
print(times)
print(len(times))
````

The above cell is creating the list and then appending to the list. If we
had created the list in a previous cell, running this cell twice would add

as you run it.

## 3. Dictionaries: the patient chart

A chart is not a list. You do not look for "the fourth item", you look for a
specific labelled field such as the consultant or the allergies. The Python
version of this is a **dictionary** (usually shortened to a **dict**). This
stores a pairing of a **key** (the label) and a **value** within a set of
curly brackets:

````python
patient = {
    "name": "Jordan Nguyen",
    "hospital_number": "RTH-102938",
    "age": 54,
    "ward": "5B",
    "bed": 12,
    "consultant": "Dr Priya Patel",
    "allergies": ["Penicillin"],
}
print(patient)
````

Each line inside the brackets is `key: value`, separated by commas. The keys
in this case are strings with the values being anything from text to numbers
to even a list as seen with `"allergies"`.

To lookup a value, provide the key in square brackets like you would an
index for a list except you're using a label not a position.

````python
print(patient["consultant"])
print(patient["allergies"])
````

Asking for a key that is not there, such as `patient["weight"]`, stops with
a `KeyError`: Python's way of saying "this chart has no field with that
label". Keys must match exactly, including capitals.

You can change the value or add a new key by using assignment:

````python
patient["consultant"] = "Dr Sam Okafor"  # care handed over
patient["diet"] = "Normal"  # a new field
print(patient["consultant"])
print(patient["diet"])
````

## 4. Loops and decisions: reviewing the observations

### Doing something to every reading

A **for loop** repeats some code once for each item in a list. You can read
`for temperature in temperatures:` as "for each temperature in the list of
temperatures, do the following". Each time round, the variable `temperature`
holds the next reading.

````python
temperatures = jordan["observations"]["temperature"]
for temperature in temperatures:
    print(temperature)
print("Done")
````

The **indented** lines (moved in by four spaces) are the body of the
loop: they run once per reading. `print("Done")` is not indented, so it runs
once, after the loop has finished. In Python, indentation is not just
layout: it decides what belongs to what. The `:` at the end of the `for`
line is required.

A common pattern is to keep a running total, starting from 0 and adding to
it each time round the loop. Here are the oral fluids Jordan drank this
morning:

````python
oral_fluids_ml = [200, 150, 250, 100]
total_ml = 0
for volume_ml in oral_fluids_ml:
    total_ml = total_ml + volume_ml
print(f"Oral intake so far: {total_ml} ml")
````

`total_ml = total_ml + volume_ml` means, "work out the old total plus this
volume and store the result as the new total". Python has a shorthand for
this as well, `total_ml += volume_ml`.

### Making a decision

Clinical reasoning is full of "if this, do that". Python writes this as
`if`, `elif` (short for 'else if'), and `else`. Python runs through each
condition and only runs the first one that is true.

Conditions are usually some form of comparison:

| Python | Meaning |
|---|---|
| `a > b`, `a < b` | greater than, less than |
| `a >= b`, `a <= b` | greater than or equal to, less than or equal to |
| `a == b` | equal to (two equals signs: one `=` stores a value) |
| `a != b` | not equal to |

Is Jordan's temperature coming down? Let's compare the latest reading to the
first:

````python
first_temperature = temperatures[0]
latest_temperature = temperatures[-1]

if latest_temperature < first_temperature:
    print("Temperature falling")
elif latest_temperature > first_temperature:
    print("Temperature rising")
else:
    print("Temperature unchanged")
````

As with loops, each condition ends with a `:` and the code that belongs to
it is indented. Only one branch ever runs.

## 5. Functions: protocols you can reuse

A protocol is a set of steps that are written once and then followed the
same way for every patient. A **function** is this idea in code. You give a
set of steps a name and then use that name whenever you need the steps.
You've already called functions that someone else wrote such as `print`,
`len`, and `sample_patient`. Now you'll write one.

````python
def calculate_bmi(weight_kg, height_m):
    return weight_kg / height_m**2


print(calculate_bmi(81.4, 1.72))  # Jordan
print(calculate_bmi(70.0, 1.75))  # another patient
````

Reading the function line by line:

- `def` starts a function definition, followed by its name.
- `weight_kg` and `height_m` in the brackets are its **parameters**: the
  information it needs. Each time you call the function, the values you
  give in the brackets are stored in those names.
- The indented lines are the steps. `return` returns the answer to wherever
  the function was called.

Running the `def` cell creates the function but prints nothing. The
function's steps only run when you call it.

`return` is not `print`. `print` merely shows a value on screen, `return`
gives the value back to be used for other code. This could be to save the
value to a variable, check its value, or to print it.
