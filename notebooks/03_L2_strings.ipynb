## Text is a value too

Putting text in quotes tells Python to treat it as a string. You can use
either single quotes `'...'` or double quotes `"..."`. Both function
identically, however, this course uses double quotes.

`+` between two strings **joins** them end to end. Programmers call this
**concatenation**. Python does not add a space for you, so you must put it
in yourself.

````python
first_name = "Jordan"
surname = "Nguyen"
full_name = first_name + " " + surname
print(full_name)
````

## f-strings: the easy way to build a message

Joining strings with a `+` is rather clumsy if the message has multiple
values that need to be inserted. A **f-string** is a string with an `f` in
front of the first quotation mark. The string can then contain any values in
`{}` which are replaced with their values.

After the name, you can include a format. `{temp_c:.1f}` means "present
`temp_c` to one decimal place". The `f` in `.1f` is for float. This just
rounds the number for presentation but doesn't affect `temp_c`.

````python
ward = "5B"
temp_c = 37.84
print(f"{full_name}, ward {ward}, bed {bed}")
print(f"Temperature {temp_c:.1f} °C")
print(f"Temperature {temp_c:.2f} °C")
````

## Methods: tools that come with every string

Text written by humans is inherently a mess. There are errant spaces, caps
lock is left on, separators are inconsistent, and more. All strings have
in-built functionality to clean them up a bit. These functions are called
**methods**. To call a method, you use a period followed by the method name
and brackets, e.g.: `name.upper()`.

| Method | What it gives back | Example result |
|---|---|---|
| `.strip()` | the text without spaces at the start and end | `"  5B "` becomes `"5B"` |
| `.lower()`, `.upper()` | all lower case, all capitals | `"Ramipril"` becomes `"ramipril"` |
| `.title()` | a capital at the start of each word | `"jordan nguyen"` becomes `"Jordan Nguyen"` |
| `.replace(old, new)` | every `old` swapped for `new` | `"12-03"` with `.replace("-", "/")` becomes `"12/03"` |
| `.startswith(text)` | `True` or `False` | `"RTH-1".startswith("RTH")` is `True` |

There is one rule that is more important than all the others. **A method
never modifies the original string.** It always returns a new string that
you can assign to a variable if you want to store it. You can also apply a
method to this new string, this is called a method chain and Python
evaluates it from left to right.

````python
allergy = "  PENICILLIN "
print(allergy.strip())
print(allergy.strip().lower())
print(allergy)  # unchanged: we never stored a new value

allergy = allergy.strip().title()  # store the tidied version
print(allergy)
print(allergy.startswith("Pen"))  # True or False, as in the basics notebook
````

## Text that looks like a number

`"12"` (in quotes) and `12` are different things to Python. The first is
a string that happens to contain digits; the second is a number. You can
ask Python which is which with `type()`.

````python
bed_as_text = "12"
bed_as_number = 12
print(type(bed_as_text))
print(type(bed_as_number))
print(type(weight_kg))
````

Read `<class 'str'>` as "This is a str", i.e., text. `int` and `float` are
the two types of number from Level 1.

Mixing the two with `+` causes an error, because Python does not know if you
mean "join" or "add":

```
print("Bed " + 12)
TypeError: can only concatenate str (not "int") to str
```

Read the last line of an error first, this says that `+` can only join a
string (`str`) to another string and you gave it a number (`int`). There are
two fixes for this depending on what you want:

- To put a number **into text** you can either use an f-string `f"Bed
  {bed}"` or convert the number to text using `str(bed)`.
- To **calculate** with text that holds a number, convert it with `int()`
  for whole numbers or `float()` for decimals.

````python
dose_text = "500"  # for example, read from a form
dose_mg = int(dose_text)
print(f"Two doses: {dose_mg * 2} mg")

height_text = "1.72"
height_m = float(height_text)
print(f"BMI: {81.4 / height_m**2:.1f}")

print("Bed " + str(bed))
````
