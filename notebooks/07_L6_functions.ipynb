# Level 6: Functions, and building NEWS2 from parts

A **function** is a named, reusable step. You've been using functions from
the start (`print`, `len`, `round`), now you'll learn to write your own.
Functions are important because they enable you to take a large problem and
break it up into small pieces, get each piece correct in isolation, and then
combine the pieces together. This process of breaking a problem into parts
is called **decomposition**. It is the central idea of this level and most
real programs.

NEWS2 is a good example, because it is already built that way. Seven
observations are each turned into a score, the scores are added up, and the
total (with one extra rule) gives a risk band. By the end of this level you
will have built that calculator from parts, then used it to make a watch
list for a whole ward:

```
one observation   ->  score_pulse, score_temperature, ...   (one score each)
all observations  ->  news2_scores  ->  news2_total  ->  news2_risk
the whole ward    ->  build_watch_list   (uses news2_total and news2_risk)
```

## Writing a function

Here is body mass index in a function:

````python
def bmi(weight_kg, height_m):
    """Body mass index in kg/m², from weight in kg and height in metres."""
    return weight_kg / height_m**2


jordan_bmi = bmi(81.4, 1.72)
print(round(jordan_bmi, 1))
print(round(bmi(60, 1.65), 1))
````

Reading this line by line:

- The `def` keyword indicates to Python that we're going to define a new
  function. We give it the name `bmi`.
- `weight_kg` and `height_m` are **parameters** of the function: they are
  names for the values that the function will be provided with. When we
  **call** the function, e.g., `bmi(81.4, 1.72)`, the values within the
  brackets (the **arguments**) are matched to the parameters in order.
- The text in triple quotes is a **docstring** which is a single line
  explaining what the function does. You can see it if you run `help(bmi)`
  in a new cell. You should write one for any function you make. Your future
  self (and anyone else reading your code) will thank you.
- `return` hands a value back to the code that called the function. That is
  how `jordan_bmi` gets its value.

Defining a function does not run it. The indented body only runs each time
the function is called.

**Return vs Print** `print` displays a value to the screen for a human to
read. `return` provides the value back to the program for it to use, such as
storing the value, comparing, or passing along. A function without a
`return` statement returns a special value called `None`. This is Python's
way of indicating "nothing".

## One observation, one function

In Level 3, you worked out how to turn a single observation into a NEWS2
score using `if`, `elif`, and `else`. What a function lets you do is wrap
that logic, give it a name, and then use it for any patient. Let's take the
respiration rate from this part of the chart:

| Respiration rate (breaths/min) | Score |
|--------------------------------|-------|
| 8 or fewer                     | 3     |
| 9-11                           | 1     |
| 12-20                          | 0     |
| 21-24                          | 2     |
| 25 or more                     | 3     |

````python
def score_respiration_rate(breaths_per_min):
    """NEWS2 score for respiration rate in breaths per minute."""
    if breaths_per_min <= 8:
        return 3
    elif breaths_per_min <= 11:
        return 1
    elif breaths_per_min <= 20:
        return 0
    elif breaths_per_min <= 24:
        return 2
    else:
        return 3


print(score_respiration_rate(16))
print(score_respiration_rate(22))
print(score_respiration_rate(8))
````

Each band is tested with `<=` to the top number in order from the lowest
band up. The first test that is true determines the score and the `return`
ends the function there. This means that `<= 11` is only reached for rates
above 8 which is exactly the 9-11 band.

## Functions that use other functions

A function can call another function. This version of BMI takes in the
height in centimetres and instead of repeating the formula, it converts the
height and passes the work on to `bmi`.

````python
def bmi_from_cm(weight_kg, height_cm):
    """Body mass index in kg/m², with height recorded in centimetres."""
    return bmi(weight_kg, height_cm / 100)


print(round(bmi_from_cm(81.4, 172), 1))
````

The formula now lives in one place. If it ever needed to change, every
function built on it would change as well.

## Failing loudly

What should a NEWS2 calculator do if the respiration rate was never
recorded? Scoring this as 0 would be quiet and wrong: a patient breathing at
30 a minute could look well because no one counted. A safer function would
stop and say what was missing. You can make a function stop using `raise`:

````python
def bmi(weight_kg, height_m):
    """Body mass index in kg/m². Raises ValueError if height is not above 0."""
    if height_m <= 0:
        raise ValueError(f"height must be in metres and above 0, got {height_m}")
    return weight_kg / height_m**2


print(round(bmi(81.4, 1.72), 1))
````

This cell defines `bmi` a second time. This silently overwrites the original
function as a new `def` is made with the same name. This means that the
`bmi_from_cm` function which calls `bmi` now uses this safer variant as
well.

Called as `bmi(81.4, 0)`, the new version stops with:

```
ValueError: height must be in metres and above 0, got 0
```

A **ValueError** is the typical error Python uses to indicate that "this
value is not something I can work with". The error message is intended for
human consumption, so make sure that it actually tells someone what went
wrong. An error that you can read is better than an answer you can't trust.

## Sorting by a field

The watch list needs to be sorted to ensure the sickest patients move to the
front of the list. The `sorted` function sorts numbers or text, however, it
can't compare two patient dictionaries. We need to tell it what to compare.
This can be provided to the `key` argument as a function that `sorted` uses
to compare each element.

````python
patients_by_age = [
    {"name": "Grace Evans", "age": 58},
    {"name": "Kai Morgan", "age": 91},
    {"name": "Isla Fraser", "age": 34},
]


def age_of(patient):
    """The patient's age, used as a sorting key."""
    return patient["age"]


print(sorted(patients_by_age, key=age_of))
print(sorted(patients_by_age, key=age_of, reverse=True))
````

Two new things here.

- `key=age_of` has no brackets after `age_of`. With brackets, `age_of()`
  would call the function straight away. Without them, you hand `sorted`
  the function itself, as a value, for it to call on each item. A function
  is a value like any other: it can be stored, passed and called later.
- `reverse=True` = largest first, items with the same key retain original
  order.

`sorted` returns a new sorted list and keeps the original list unchanged.

## Before you move on

- `def name(parameters):` defines a function; calling `name(arguments)`
  runs it.
- `return` gives a value back, `print` just shows it. A function without a
  `return` gives `None`.
- All functions should have a one line docstring indicating what it returns.
- Build big functions from small, checked ones, and test each on its
  boundaries.
- If there's a missing or impossible input, `raise ValueError` with a
  message indicating what's wrong rather than silently guessing.
